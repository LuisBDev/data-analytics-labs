### 2.2. División de train - test y Balanceo de datos

#### Interpolation Method

---
* Train - Test

```python
df_Interpolation_Method
```

```python
'''
la división del los datos será de un 70 - 30
'''
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df_Interpolation_Method.drop(columns=['FLAG']) #Representando las X
                                                    , df_Interpolation_Method['FLAG'] #Representando la variable y
                                                    , test_size=0.3
                                                    , random_state= semilla
                                                    , stratify=df_Interpolation_Method['FLAG']
                                                   )
```

```python
y_train.value_counts()
```

```python
y_test.value_counts()
```

```python
%%time
pd.concat([y_test ,X_test], axis=1).to_csv('{}Interpolation_Method_Test.csv'.format(ruta_df), index=False)
```

---
* Balanceo de datos

```python
from imblearn.over_sampling import ADASYN 
ada = ADASYN(random_state= semilla
             , n_neighbors= 20
             , sampling_strategy=.8
             )
```

```python
X, y = ada.fit_resample(X_train.drop(columns=['CONS_NO']), y_train)
```

```python
%%time
pd.concat([y, X], axis=1).to_csv('{}Interpolation_Method_ADASYN_Train.csv'.format(ruta_df), index=False)
```

```python
from imblearn.over_sampling import SMOTE 
smote = SMOTE(random_state= semilla
             , sampling_strategy=.8
             )
```

```python
X, y = smote.fit_resample(X_train.drop(columns=['CONS_NO']), y_train)
```

```python
%%time
pd.concat([y, X], axis=1).to_csv('{}Interpolation_Method_SMOTE_Train.csv'.format(ruta_df), index=False)
```
