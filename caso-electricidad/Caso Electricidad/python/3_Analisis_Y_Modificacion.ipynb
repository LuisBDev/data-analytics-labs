### 3.1. Interpolation Method

* ADASYN

```python
df_IM_A = pd.read_csv('{}Interpolation_Method_ADASYN_Train.csv'.format(ruta_df))
```

```python
df_IM_A.describe()
```

```python
from ipywidgets import interact, interact_manual
df_IM_ADASYN_grafico = Graficos(df_IM_A.drop(df_IM_A.filter(like='/', axis=1).columns, axis=1).drop(columns='Three_Sigma_Rule'))
```

```python
interact_manual(
    df_IM_ADASYN_grafico.Correlacion
    , method = ['pearson', 'kendall', 'spearman']
    , TipoGrafico = np.arange(3)
)
```

```python
interact_manual(
    df_IM_ADASYN_grafico.BoxPlot
    , Caracteristica = df_IM_A.loc[:, 'Sum':]
    , limite = ['Ninguno', 'FLAG = 0', 'FLAG = 1', 'General']
)
```

```python
#_temp = df_IM_A.query("Mean < @_filtro")
_temp = df_IM_A
_temp_grafico = Graficos(_temp.drop(df_IM_A.filter(like='/', axis=1).columns, axis=1).drop(columns='Three_Sigma_Rule'))
_sel_col = _temp_grafico.calcular_correlacion('FLAG', 5, "alta corr") # "alta corr", "baja corr"
_sel_col.append("FLAG")
_sel_col
```

```python
#Exportación del dataset modificado 2 
df_IM_A[_sel_col].to_csv('{}Interpolation_Method_ADASYN_Train_Modificado.csv'.format(ruta_df), index=False)
```
