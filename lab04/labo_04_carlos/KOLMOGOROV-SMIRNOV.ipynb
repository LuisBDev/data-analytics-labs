```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import year

# Crear sesión de Spark
spark = SparkSession.builder \
    .appName("KolmogorovSmirnovTest") \
    .getOrCreate()

# Cargar datos
merge_pyspark = spark.read.parquet("merge_pyspark")

```

```python
# Extraer año de la fecha
merge_pyspark = merge_pyspark.withColumn("year", year("Fecha"))

# Conjuntos
conjunto1 = merge_pyspark.filter((merge_pyspark.year >= 2018) & (merge_pyspark.year <= 2019))
conjunto2 = merge_pyspark.filter(merge_pyspark.year == 2020)
```

```python
from scipy.stats import ks_2samp

c1_pd = conjunto1.select("price").toPandas()["price"].dropna()
c2_pd = conjunto2.select("price").toPandas()["price"].dropna()

ks_stat, ks_pvalue = ks_2samp(c1_pd, c2_pd)
print("Kolmogorov-Smirnov Test - PRICE")
print("KS estadístico:", ks_stat)
print("p-value:", ks_pvalue)
```

```python
c1_pd = conjunto1.select("FN").toPandas()["FN"].dropna()
c2_pd = conjunto2.select("FN").toPandas()["FN"].dropna()

ks_stat, ks_pvalue = ks_2samp(c1_pd, c2_pd)
print("Kolmogorov-Smirnov Test - FN")
print("KS estadístico:", ks_stat)
print("p-value:", ks_pvalue)
```

```python
from scipy.stats import ks_2samp
c1_pd = conjunto1.select("Active").toPandas()["Active"].dropna()
c2_pd = conjunto2.select("Active").toPandas()["Active"].dropna()

ks_stat, ks_pvalue = ks_2samp(c1_pd, c2_pd)
print("Kolmogorov-Smirnov Test - ACTIVE")
print("KS estadístico:", ks_stat)
print("p-value:", ks_pvalue)
```

```python
c1_pd = conjunto1.select("age").toPandas()["age"].dropna()
c2_pd = conjunto2.select("age").toPandas()["age"].dropna()

ks_stat, ks_pvalue = ks_2samp(c1_pd, c2_pd)
print("Kolmogorov-Smirnov Test - AGE")
print("KS estadístico:", ks_stat)
print("p-value:", ks_pvalue)
```

```python
from scipy.stats import ks_2samp
import pandas as pd

# Columnas numéricas a comparar
columnas = ["price", "FN", "Active", "age"]

resultados = []

for col in columnas:
    # Extraer datos de Spark → Pandas y eliminar nulos
    c1_pd = conjunto1.select(col).toPandas()[col].dropna()
    c2_pd = conjunto2.select(col).toPandas()[col].dropna()
    
    # KS-2Sample Test
    ks_stat, ks_pvalue = ks_2samp(c1_pd, c2_pd)
    
    resultados.append({
        "Variable": col,
        "KS Estadístico": ks_stat,
        "p-value": ks_pvalue
    })

# Convertir a tabla
tabla_resultados = pd.DataFrame(resultados)

print(tabla_resultados)
```
